```python
import backtrader as bt
import pandas as pd
import numpy as np
from datetime import datetime
import yfinance as yf
import matplotlib.pyplot as plt

# Custom Strategy Class
class LongTrendHighMomentum(bt.Strategy):
    params = dict(
        risk_percent=2,  # Risk percentage per position
        max_position_percent=10,  # Maximum position size in percentage
        max_positions=10,  # Max number of positions
        dollar_volume_threshold=50_000_000,  # Average dollar volume threshold
        stock_price_threshold=5,  # Minimum stock price
        atr_multiplier=5,  # Stop loss multiplier
        trailing_stop_percent=0.25,  # Trailing stop percentage
    )

    def __init__(self):
        self.stop_dict = {}
        self.size_dict = {}
        self.indicators = {}

        # Initialize lists to store portfolio values and dates
        self.portfolio_values = []
        self.dates = []

        # S&P 500 index indicators
        index = self.datas[0]  # Assuming S&P 500 index data is the first data feed
        self.index_sma_100 = bt.indicators.SMA(index.close, period=100)

        # Indicators for each stock
        for d in self.datas[1:]:
            self.indicators[d] = {}
            self.indicators[d]['sma_25'] = bt.indicators.SMA(d.close, period=25)
            self.indicators[d]['sma_50'] = bt.indicators.SMA(d.close, period=50)
            self.indicators[d]['roc_200'] = bt.indicators.RateOfChange(d.close, period=200)
            self.indicators[d]['atr'] = bt.indicators.ATR(d, period=20)
            self.indicators[d]['avg_dollar_volume'] = bt.indicators.SimpleMovingAverage(
                d.volume * d.close, period=20
            )

    def next(self):
        # Record the current portfolio value and date
        self.portfolio_values.append(self.broker.getvalue())
        self.dates.append(self.datas[0].datetime.date(0))

        # Ensure index SMA is ready
        if len(self.datas[0]) < 100:
            return

        # Check index filter
        if self.datas[0].close[0] < self.index_sma_100[0]:
            return

        # Collect potential trades
        potential_trades = []
        for d in self.datas[1:]:
            # Ensure indicators are ready
            if len(d) < 200:
                continue  # Skip until indicators are ready

            ind = self.indicators[d]

            # Filters
            if ind['avg_dollar_volume'][0] < self.p.dollar_volume_threshold:
                continue
            if d.close[0] < self.p.stock_price_threshold:
                continue
            if ind['sma_25'][0] <= ind['sma_50'][0]:
                continue

            # Add to potential trades
            potential_trades.append((ind['roc_200'][0], d))

        # Sort by highest ROC over the last 200 days
        potential_trades.sort(key=lambda x: x[0], reverse=True)

        # Limit to max positions
        open_positions = len([p for p in self.datas[1:] if self.getposition(p).size > 0])
        positions_needed = self.p.max_positions - open_positions
        trades_to_take = potential_trades[:positions_needed]

        # Enter trades
        for roc, d in trades_to_take:
            if not self.getposition(d):
                # Calculate position size
                cash = self.broker.get_cash()
                max_trade_value = cash * (self.p.max_position_percent / 100)
                size = int(max_trade_value // d.close[0])

                if size > 0:
                    self.buy(data=d, size=size)
                    # Stop loss price
                    stop_price = d.close[0] - self.p.atr_multiplier * self.indicators[d]['atr'][0]
                    self.stop_dict[d] = stop_price
                    self.size_dict[d] = size

        # Manage open positions
        for d in self.datas[1:]:
            position = self.getposition(d)
            if position.size > 0:
                ind = self.indicators[d]
                # Check stop loss
                if d.close[0] <= self.stop_dict.get(d, 0):
                    self.close(data=d)
                    del self.stop_dict[d]
                    del self.size_dict[d]
                else:
                    # Update trailing stop
                    trailing_stop_price = d.close[0] * (1 - self.p.trailing_stop_percent)
                    if trailing_stop_price > self.stop_dict.get(d, 0):
                        self.stop_dict[d] = trailing_stop_price

    def notify_order(self, order):
        if order.status in [order.Completed]:
            if order.isbuy():
                print(f'BUY EXECUTED: {order.data._name} Price: {order.executed.price}')
            elif order.issell():
                print(f'SELL EXECUTED: {order.data._name} Price: {order.executed.price}')

    def notify_trade(self, trade):
        if trade.isclosed:
            print(f'TRADE CLOSED: {trade.data._name} Profit: {trade.pnl}')

if __name__ == '__main__':
    import backtrader as bt
    import pandas as pd
    from datetime import datetime
    import yfinance as yf
    import matplotlib.pyplot as plt

    # Set the start and end dates
    start_date = '1990-01-01'
    end_date = '2023-10-01'  # Use the latest date with available data

    # Get the list of S&P 500 tickers from Wikipedia
    url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
    tables = pd.read_html(url)
    sp500_table = tables[0]
    tickers = sp500_table['Symbol'].tolist()
    tickers = [ticker.replace('.', '-') for ticker in tickers]  # Adjust tickers for Yahoo Finance

    # Initialize Cerebro engine
    cerebro = bt.Cerebro()

    # Add Strategy
    cerebro.addstrategy(LongTrendHighMomentum)

    # Add S&P 500 index data
    index_df = yf.download('^GSPC', start=start_date, end=end_date)
    index_data = bt.feeds.PandasData(dataname=index_df)
    cerebro.adddata(index_data, name='S&P 500 Index')

    # Load data feeds for each ticker
    for ticker in tickers:
        try:
            df = yf.download(ticker, start=start_date, end=end_date)
            if not df.empty and len(df) >= 200:
                data = bt.feeds.PandasData(dataname=df)
                cerebro.adddata(data, name=ticker)
            else:
                print(f'Skipping {ticker}: insufficient data')
        except Exception as e:
            print(f'Error downloading data for {ticker}: {e}')

    # Set starting cash
    cerebro.broker.set_cash(100000)

    # Set commission
    cerebro.broker.setcommission(commission=0.001)

    # Run backtest
    print(f'Starting Portfolio Value: {cerebro.broker.getvalue():.2f}')
    results = cerebro.run()
    print(f'Ending Portfolio Value: {cerebro.broker.getvalue():.2f}')

    # Retrieve the strategy instance
    strat = results[0]

    # Create a pandas Series for portfolio values with proper dates
    portfolio_values = pd.Series(strat.portfolio_values, index=pd.to_datetime(strat.dates))

    # Plot the equity curve
    portfolio_values.plot(figsize=(12, 6))
    plt.title('Equity Curve')
    plt.xlabel('Date')
    plt.ylabel('Portfolio Value')
    plt.show()
```
